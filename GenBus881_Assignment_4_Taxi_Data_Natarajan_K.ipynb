#Data Analysis for a taxi Company
##Determine if there is a significant difference in the average tip when customers take a yellow cab versus a green cab.  (There are two different colors of taxis in New York.)  We have data with ride information. We will use 2-sample t-test to perform the evaluation.

```python
# Import the appropriate Python libraries.
import pandas as pd
import seaborn as sns
from scipy import stats
```

Import the Data

```python
# Load the CSV file into a dataframe.
df_taxi = pd.read_csv('https://bit.ly/taxi-samples')
```

#Segment the data
Create one dataframe with yellow cab data and another with green cab data.  and call them "df_taxi_yellow" and "df_taxi_green"

```python
# Select rows that meet certain criteria.
df_taxi_yellow = df_taxi[df_taxi.color == 'yellow']
df_taxi_green = df_taxi[df_taxi.color == 'green']
```

Generate a histogram to visualize the distribution of tips within the yellow cab data.

```python
# Plot the distribution of ratings for yellow taxis.
sns.displot(df_taxi_yellow['tip']);
```

Generate a histogram to visualize the distribution of tips within the green cab data.

```python
# Plot the distribution of ratings for green taxis.
sns.displot(df_taxi_green['tip']);
```

#Perform a Normality Test

Prior to performing a hypothesis test, we perform a normality test. It helps us understand what hypothesis test would be appropriate. After conducting this test, if the p-value is greater than 0.05, the data is normal enough for our purposes.

```python
# Put tip field in each dataframe into its own variable.
tip_yellow = df_taxi_yellow['tip']
tip_green = df_taxi_green['tip']
```

```python
# Conduct a normality test.
print(stats.normaltest(tip_yellow))
print(stats.normaltest(tip_green))
```

Based on the normality test, the tip_yellow data is normal enough for our purposes.

Prior to conducting the 2-sample t-test, perform a Bartlett's test to evaluate the equal variance of the two datasets. If p-value is > 0.05, we assume no differences in variance.


```python
# Perform a Bartlett's test to evaluate equal variance of the two datasets.
stats.bartlett(tip_yellow,tip_green)
```

### Perform a 2-sample t-test between the two sets of tip data

When we conduct this hypothesis test, if the p-value is > 0.05, we "fail to reject the null" and assume no differences between the two samples.  When the p-value is < 0.05, we "reject the null" and state there is a difference.

Based on the P Value of 0.964, we "fail to reject the null" - we have to assume there are no differences between the two samples.

```python
# Conduct a 2-sample t-test.
stats.ttest_ind(tip_yellow, tip_green)
```
